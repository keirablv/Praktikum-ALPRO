```python
import os

os.makedirs("saving_project", exist_ok=True)

print("Folder saving_project berhasil dibuat!")
```

```python
%%writefile saving_project/interest.py

def compound_interest(principal, rate, time):
    future_value = principal * (1 + rate) ** time
    return future_value

def simple_interest(principal, rate, time):
    future_value = principal + (principal * rate * time)
    return future_value
```

```python
%%writefile saving_project/display.py

def print_result(principal, rate, years, ci, si):
    print("\n===== HASIL PERHITUNGAN TABUNGAN =====")
    print(f"Saldo Awal              : ¥{principal:,.2f}")
    print(f"Bunga Tahunan           : {rate * 100:.1f}%")
    print(f"Lama Menabung           : {years} tahun")
    print("--------------------------------------")
    print(f"Nilai Akhir (Bunga Majemuk)   : ¥{ci:,.2f}")
    print(f"Nilai Akhir (Bunga Sederhana) : ¥{si:,.2f}")
    print("======================================")
```

```python
%%writefile saving_project/main.py

import sys
sys.path.insert(0, "saving_project")

from interest import compound_interest, simple_interest
from display import print_result

print("=== Program Kalkulator Bunga Tabungan ===")

principal = float(input("Masukkan saldo awal (contoh: 80000): "))
rate = float(input("Masukkan bunga tahunan (contoh: 0.05): "))
years = int(input("Masukkan lama menabung (tahun): "))

if rate > 1:
    rate = rate / 100

ci = compound_interest(principal, rate, years)
si = simple_interest(principal, rate, years)

print_result(principal, rate, years, ci, si)
```
