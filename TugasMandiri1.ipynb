**Praktikum: Toko Dorayaki Doraemon**
Program yang membantu Toko dorayaki doraemon untuk membuat nama toko, menghitung total modal awal,
menhitung total pendapatan, serta menghitung total keuntungan.

```python
print("Memilih nama toko harus berupa 3 kata, yaitu nama makanan, nama karakter, & Kata sifat")
NamaMakanan = input("Nama Makanan : ") #Agar pengguna dapat menginput nama makanan
NamaKarakter = input("Nama Karakter : ") #Agar pengguna dapat menginput nama karakter
KataSifat = input("Kata Sifat : ") #Agar pengguna dapat menginput kata sifat
print('Nama Toko anda adalah : ', NamaMakanan, NamaKarakter, KataSifat) #Agar ketiga kata di atas dapat tersimpan & terlihat menjadi satu kalimat oleh pengguna
```

```python
doraemon = 555000 
nobita = 333000
shizuka = 222000
Total_Modal = doraemon + nobita + shizuka #Agar jumlah modal dari setiap orang dapat terhitung
print('Total modal yang terkumpul :', Total_Modal) #Agar pengguna dapat mengetahui hasil dari perhitungannya
```

```python
jumlah_dorayaki = 350
harga_dorayaki = 5000
total_pendapatan = jumlah_dorayaki * harga_dorayaki #
print('Total Pendapatan pada bulan pertama : ', total_pendapatan)
```

```python
modal_awal = 1110000
pendapatan = 1750000
keuntungan = pendapatan - modal_awal #Agar keuntungan pada bulan pertama dapat terhitung
print('Total keuntungan pada bulan pertama : ', keuntungan) #Agar pengguna dapat mengetahui hasil dari perhitungannya
```
