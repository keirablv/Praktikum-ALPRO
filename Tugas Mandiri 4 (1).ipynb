**Smart Vehicle Insurance Analyzer**

```python
usia_pengemudi = int(input("Usia pengemudi: "))
jenis_kendaraan = input("Jenis kendaraan (mobil/motor): ")
nilai_kendaraan = float(input("Nilai kendaraan (Rp): "))
area_penggunaan = input("Area penggunaan (perkotaan/pedesaan): ")
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ")

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko = "tinggi"
else:
    risiko = "rendah"

if jenis_kendaraan == "motor":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"
elif jenis_kendaraan == "mobil":
     pass


if area_penggunaan == "perkotaan":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"

if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"

if risiko == "tinggi":
    persentase_premi = 0.10
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi = persentase_premi * nilai_kendaraan

pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon."
)

print(" ")
print("===== HASIL ANALISIS ASURANSI =====")
print("Usia pengemudi       :", usia_pengemudi, "tahun")
print("Jenis kendaraan      :", jenis_kendaraan)
print("Nilai kendaraan      : Rp{:,.0f}".format(nilai_kendaraan))
print("Area penggunaan      :", area_penggunaan)
print("Sistem pengaman      :", sistem_pengaman)
print("Kategori risiko      :", risiko)
print("Persentase premi     :", persentase_premi * 100, "%")
print("Premi asuransi       : Rp{:,.0f}".format(premi))
print("Pesan Doraemon       :", pesan)
```
